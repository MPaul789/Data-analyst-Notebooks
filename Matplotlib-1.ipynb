# Why Should We Go Beyond Pyplot in Matplotlib?


## 3. Why Should We Go Beyond Pyplot?

There are several reasons to use the object-oriented interface instead of relying entirely on `pyplot`.

### A. Better Control Over Multiple Figures

When several figures are open, pyplot functions operate on the current figure and its current Axes. The object-oriented interface allows us to specify exactly which Axes we want to modify.

### B. Easier Management of Subplots

When creating dashboards or visualizations containing multiple subplots, each Axes object can be customized independently.

### C. Improved Readability

Using explicit objects such as `fig`, `ax1`, and `ax2` makes it easier to understand which plot is being modified.

### D. Better Maintainability

For large projects, the object-oriented approach makes code easier to organize, debug, reuse, and extend.

## 4. What Happens to `plt.title()` When Three Figures Are Open?

Consider the following example:

```python
import matplotlib.pyplot as plt

plt.figure(1)
plt.plot([1, 2, 3], [10, 20, 30])
plt.title("Figure 1")

plt.figure(2)
plt.plot([1, 2, 3], [20, 30, 40])
plt.title("Figure 2")

plt.figure(3)
plt.plot([1, 2, 3], [30, 40, 50])
plt.title("Figure 3")
```

In this example, we create three separate figures.

- `plt.figure(1)` activates Figure 1.
- `plt.figure(2)` activates Figure 2.
- `plt.figure(3)` activates Figure 3.

Whenever we activate a figure, Matplotlib makes that figure current. If the figure already contains Axes, its current Axes becomes the target of pyplot functions.

Therefore, when we execute:

```python
plt.title("New Title")
```

the title is applied to the current Axes, which belongs to Figure 3 in this example.

**Important:** `plt.title()` does not automatically update all three figures. It modifies the title of the current Axes.

If we want to change the title of Figure 1, we can activate Figure 1 first:

```python
plt.figure(1)
plt.title("Updated Figure 1")
```

This works, but repeatedly switching between figures can become inconvenient in more complex programs.

## 5. Understanding the Figure and Axes Objects

To understand why the object-oriented approach is useful, we need to distinguish between a Figure and an Axes.

- **Figure:** The overall container that holds the plot or plots.
- **Axes:** The individual plotting area containing the data, title, axis labels, and ticks.

A single Figure can contain multiple Axes, such as when using subplots.

For example:

```python
fig, ax = plt.subplots()
```

Here:

- `fig` represents the Figure object.
- `ax` represents the Axes object.

We can then customize the Axes directly:

```python
ax.plot([1, 2, 3], [10, 20, 30])
ax.set_title("My First Plot")
ax.set_xlabel("X-axis")
ax.set_ylabel("Y-axis")
```

Unlike `plt.title()`, `ax.set_title()` explicitly identifies which Axes should receive the title.

## 6. Object-Oriented Approach With Three Figures

The following example creates three figures and gives us independent control over each one.

```python
import matplotlib.pyplot as plt

fig1, ax1 = plt.subplots()
ax1.plot([1, 2, 3], [10, 20, 30])
ax1.set_title("Figure 1")

fig2, ax2 = plt.subplots()
ax2.plot([1, 2, 3], [20, 30, 40])
ax2.set_title("Figure 2")

fig3, ax3 = plt.subplots()
ax3.plot([1, 2, 3], [30, 40, 50])
ax3.set_title("Figure 3")

plt.show()
```

In this example:

- `fig1`, `fig2`, and `fig3` represent three independent figures.
- `ax1`, `ax2`, and `ax3` represent their respective Axes.
- `ax1.set_title()` modifies the title of the first Axes.
- `ax2.set_title()` modifies the title of the second Axes.
- `ax3.set_title()` modifies the title of the third Axes.

The result does not depend on which figure was most recently activated.

## 7. Comparison Between Pyplot and the Object-Oriented Approach

| Feature | Pyplot | Object-Oriented API |
|---|---|---|
| Plot creation | `plt.plot()` | `ax.plot()` |
| Setting titles | `plt.title()` | `ax.set_title()` |
| Setting labels | `plt.xlabel()` | `ax.set_xlabel()` |
| Managing multiple figures | Uses the current figure and Axes | Explicitly references the required objects |
| Managing subplots | Can become less convenient | Provides direct control over each Axes |
| Best suited for | Quick, simple plots | Complex plots and reusable visualization code |

# GridSpec in Matplotlib

## 1. Introduction

`GridSpec` is a feature in Matplotlib that allows us to create flexible and customizable subplot layouts.

When we use `plt.subplots()`, we can create multiple plots arranged in rows and columns. However, these plots generally occupy equal-sized grid cells by default.

Sometimes, we need a more flexible layout, such as:
- One large plot on the left and two smaller plots on the right.
- One wide plot at the top and two smaller plots at the bottom.
- Multiple plots with different widths and heights.

GridSpec helps us achieve these layouts by allowing a subplot to occupy one or more rows and columns.

## 2. What Is GridSpec?

`GridSpec` defines a grid of rows and columns that determines where subplots can be placed within a figure.

It does not draw charts by itself. Instead, it defines the layout, and we create Axes objects in the grid positions.

The main components are:

- **Figure:** The overall container that holds the visualizations.
- **GridSpec:** Defines the grid structure and relative sizes of its rows and columns.
- **Axes:** The individual plotting area where data is visualized.

## 3. Syntax of GridSpec

The general syntax is:

`gridspec.GridSpec(nrows, ncols, figure=fig)`

Here:

- `nrows` specifies the number of rows.
- `ncols` specifies the number of columns.
- `figure` specifies the Figure associated with the grid.

For a grid with two rows and two columns:

gs[0, 0] selects the first row and first column.

gs[0, 1] selects the first row and second column.

gs[1, 0] selects the second row and first column.

gs[1, 1] selects the second row and second column.

GridSpec also supports slicing:

gs[:, 0] selects all rows in the first column.

gs[:, 1] selects all rows in the second column.

gs[0, :] selects all columns in the first row.

gs[1, :] selects all columns in the second row.

The colon (:) means all positions along that dimension.



# Twin Axes in Matplotlib — Two Scales, One Plot

## 1. Introduction

Twin axes in Matplotlib allow us to display two different datasets in the same plotting area, even when the datasets use different units or scales.

For example, we might want to visualize:
- Temperature in degrees Celsius and humidity in percentage.
- Sales revenue and profit margin.
- Stock price and trading volume.
- Number of hospital admissions and average length of stay.

When two quantities have different units, using a single y-axis may make one dataset difficult to interpret. Twin axes solve this problem by providing separate axes for the two datasets.

## 2. Types of Twin Axes

Matplotlib provides two main methods for creating twin axes.

### A. `ax.twinx()`

The `twinx()` method creates a second Axes that shares the original x-axis but has an independent y-axis.

- The original y-axis appears on the left.
- The second y-axis appears on the right.
- Both datasets use the same x-values or x-axis coordinates.

This is useful when comparing two quantities measured in different units.

### B. `ax.twiny()`

The `twiny()` method creates a second Axes that shares the original y-axis but has an independent x-axis.

- The original x-axis appears at the bottom.
- The second x-axis typically appears at the top.
- Both datasets use the same y-values or y-axis coordinates.

This is useful when comparing different horizontal scales against the same vertical values.



## What Is a Secondary Axis?

A secondary axis is useful when one scale can be converted into another through a known mathematical relationship.

For example, Celsius and Fahrenheit represent the same temperature using different units.

Matplotlib's `secondary_yaxis()` and `secondary_xaxis()` methods can display converted units.

Unlike twin axes, which allow independent datasets and scales, a secondary axis is intended to represent a transformation of the primary axis.




### Twin axes using twinx()

````python
import matplotlib.pyplot as plt
# Sample data
months = ["Jan", "Feb", "Mar", "Apr", "May"]
sales = [100, 150, 130, 180, 220]
temperature = [20, 25, 23, 28, 32]
# Create the first axes
fig, ax1 = plt.subplots(figsize=(9, 5))
# Plot sales on the left y-axis
line1 = ax1.plot(
    months, sales,
    color="blue",
    marker="o",
    label="Sales"
)
ax1.set_xlabel("Month")
ax1.set_ylabel("Sales (units)", color="blue")
ax1.tick_params(axis="y", labelcolor="blue")
# Create the second y-axis
ax2 = ax1.twinx()
# Plot temperature on the right y-axis
line2 = ax2.plot(
    months, temperature,
    color="red",
    marker="s",
    label="Temperature"
)
ax2.set_ylabel("Temperature (°C)", color="red")
ax2.tick_params(axis="y", labelcolor="red")
# Combine both legends
lines = line1 + line2
labels = [line.get_label() for line in lines]
ax1.legend(lines, labels, loc="upper left")
plt.title("Monthly Sales and Temperature")
fig.tight_layout()
plt.show()
````

## Secondary Axis in Matplotlib

A secondary axis is an additional axis that displays a different scale or unit
corresponding to the primary axis.

Matplotlib provides two methods:

1. `secondary_xaxis()` - Creates a secondary x-axis, usually at the top.
2. `secondary_yaxis()` - Creates a secondary y-axis, usually on the right.

Unlike `twiny()` and `twinx()`, secondary axes are designed for values that
can be converted from the primary axis using mathematical functions.

For example, temperature can be displayed in Celsius on the primary y-axis
and Fahrenheit on the secondary y-axis.

The conversion formulas are:

Fahrenheit = (Celsius * 9/5) + 32

Celsius = (Fahrenheit - 32) * 5/9

````python
import matplotlib.pyplot as plt  # Import Matplotlib
# Define temperature values in Celsius
hours = [1, 2, 3, 4, 5]
celsius = [0, 10, 20, 30, 40]
# Create a figure and primary Axes
fig, ax = plt.subplots(figsize=(8, 5))
# Plot temperature against time using Celsius
ax.plot(hours, celsius, marker='o', color='blue')
# Label the primary x-axis and y-axis
ax.set_xlabel("Time (hours)")
ax.set_ylabel("Temperature (°C)", color='blue')
# Color the primary y-axis tick labels blue
ax.tick_params(axis='y', labelcolor='blue')
# Define a function to convert Celsius to Fahrenheit
def celsius_to_fahrenheit(c):
    return (c * 9/5) + 32

# Define a function to convert Fahrenheit to Celsius
def fahrenheit_to_celsius(f):
    return (f - 32) * 5/9

# Create a secondary y-axis on the right
# Matplotlib uses these functions to convert between the two scales
secax = ax.secondary_yaxis(
    'right',
    functions=(celsius_to_fahrenheit, fahrenheit_to_celsius)
)

# Label the secondary y-axis
secax.set_ylabel("Temperature (°F)", color='red')

# Color the secondary y-axis tick labels red
secax.tick_params(axis='y', labelcolor='red')

# Add a title
ax.set_title("Temperature in Celsius and Fahrenheit")
# Adjust the layout
fig.tight_layout()
# Display the plot
plt.show()
````

# Log and Symlog Scales in Matplotlib

## 1. Understanding a Normal (Linear) Scale

A normal graph uses a **linear scale** by default.

For example, consider the following values:

```python
y = [10, 20, 30, 40, 50]
```

The difference between consecutive values is always 10.

On a linear scale, equal differences in values have equal spacing on the axis.

For example:

- The difference between 10 and 20 is 10.
- The difference between 20 and 30 is 10.
- The difference between 30 and 40 is 10.

This works well when the data values are within a reasonable range.

However, suppose the data is:

```python
y = [1, 10, 100, 1000, 10000]
```

The values increase very quickly. When the data contains very small and very large values, a linear scale can make the smaller values difficult to visualize.

To solve this problem, Matplotlib provides logarithmic scales.

## 2. What Is a Logarithmic Scale (`log`)?

A **logarithmic scale** helps us visualize data that spans a wide range of values.

Instead of spacing values according to their numerical differences, a base-10 logarithmic scale spaces powers of 10 equally.

For example:

| Value | Power of 10 |
|---:|---:|
| 1 | \(10^0\) |
| 10 | \(10^1\) |
| 100 | \(10^2\) |
| 1,000 | \(10^3\) |
| 10,000 | \(10^4\) |

Each value is ten times the previous value, but consecutive powers of 10 appear equally spaced on a base-10 logarithmic axis.


**Important:** A standard logarithmic axis cannot display zero or negative values because their real logarithms are undefined.

## 3. What Is a Symmetric Logarithmic Scale (`symlog`)?

A **symmetric logarithmic scale**, also called `symlog`, is useful when the dataset contains negative values, zero, and positive values.

Consider the following dataset:

```python
y = [-1000, -100, -10, -1, 0, 1, 10, 100, 1000]
```

This dataset contains:

- Negative values: `-1000`, `-100`, `-10`, `-1`
- Zero: `0`
- Positive values: `1`, `10`, `100`, `1000`

A standard logarithmic scale cannot represent all these values, but `symlog` can.

### How Does `symlog` Work?

It combines two types of scaling:

1. **Near zero:** It uses a linear scale.
2. **Far from zero:** It uses logarithmic scaling on both the positive and negative sides.

The parameter `linthresh` controls the approximate boundary of the linear region around zero.

For example:

```python
ax.set_yscale('symlog', linthresh=1)
```

Here, `linthresh=1` specifies the boundary of the linear region around zero. Values between -1 and 1 are in the linear region, while values farther away transition to logarithmic scaling.

## Conclusion

- Use `log` when the data is positive and spans a wide range of magnitudes.
- Use `symlog` when the data contains negative values, zero, and positive values.
- The main difference is that `symlog` includes a linear region around zero, whereas a standard logarithmic scale cannot represent zero or negative values.

````python
import matplotlib.pyplot as plt

# Create x-axis values
x = [1, 2, 3, 4, 5]

# Create y-axis values that increase by a factor of 10
y = [1, 10, 100, 1000, 10000]

# Create a figure and Axes
fig, ax = plt.subplots()

# Draw a line graph with circular markers
ax.plot(x, y, marker='o')

# Change the y-axis to a logarithmic scale
ax.set_yscale('log')

# Add a title and axis labels
ax.set_title("Logarithmic Scale")
ax.set_xlabel("X values")
ax.set_ylabel("Y values")

# Display the graph
plt.show()
````

# Custom Legends in Matplotlib

## 1. Introduction

In Matplotlib, a **legend** explains what different lines, markers, colors, or shaded regions represent in a chart. By default, Matplotlib creates a legend using the labels provided when plotting data.

However, sometimes the default legend is insufficient. For example, you may want to combine entries from multiple axes, customize the legend labels, display a legend outside the plot, or represent a shaded region that does not have a standard legend entry.

**Custom legends** allow you to control exactly what appears in the legend and how it is displayed.

## 2. Handles and Labels

A legend consists of two main components:

- **Handles:** The visual objects displayed in the legend, such as lines, markers, or colored patches.
- **Labels:** The text that describes each handle.

For example, a blue line might have the label `Sales`, while a red line might have the label `Profit`.

You can provide handles and labels explicitly to control the legend entries.

```python
ax.legend(handles, labels)
```

The number and order of handles should match the number and order of labels.

## 3. Retrieving Handles and Labels

Matplotlib provides the following method to retrieve legend handles and labels automatically:

```python
handles, labels = ax.get_legend_handles_labels()
```

- `handles` contains the plotted objects eligible for the legend.
- `labels` contains their corresponding labels.

This is useful when you want to modify the default legend or combine entries from multiple axes.

## 4. Building Legend Entries Manually

You can create legend entries manually when you need complete control over the legend.

For example, you can create a line handle using `Line2D` and a colored rectangular handle using `Patch`.

- **`Line2D`** represents a line or marker in the legend.
- **`Patch`** represents a filled shape, such as a rectangle.

These objects can act as *proxy artists*. A proxy artist is created specifically for the legend and does not need to represent an object actually plotted in the chart.

Proxy artists are useful for representing shaded regions, categories, thresholds, or other visual elements that do not have a suitable default legend entry.

## 5. Combining Entries from Multiple Axes

Sometimes a chart contains multiple axes, such as a primary y-axis and a secondary y-axis.

Each axis may have its own plotted objects and legend entries. To display all entries in one legend, retrieve the handles and labels from both axes and combine them.

This creates a single, cohesive legend for the entire chart.

## 6. Placing a Legend Outside the Axes

A legend placed inside a chart can cover important data points or lines.

The `bbox_to_anchor` parameter lets you position the legend relative to the axes. Combined with `loc`, it can place the legend outside the plotting area.

For example:

```python
ax.legend(loc="upper left", bbox_to_anchor=(1.02, 1))
```

Here:

- `loc="upper left"` selects the legend's upper-left corner as the reference point.
- `bbox_to_anchor=(1.02, 1)` positions that reference point slightly to the right of the axes' upper-right corner.

You may need to adjust the figure layout or use `fig.tight_layout()` so the outside legend is not clipped.

## 7. Key Functions and Classes

| Function or class | Purpose |
|---|---|
| `ax.legend(handles, labels)` | Creates a legend using explicitly supplied handles and labels. |
| `ax.get_legend_handles_labels()` | Retrieves eligible legend handles and their labels from an axis. |
| `Line2D` | Creates a line or marker object, including one used only in a legend. |
| `Patch` | Creates a filled shape that can be used as a legend handle. |
| `bbox_to_anchor` | Controls the legend's anchor position relative to the axes or figure. |
| `loc` | Specifies which point of the legend aligns with its anchor. |

## 8. Summary

Custom legends are useful when the default legend does not meet your requirements. You can specify handles and labels manually, combine entries from different axes, create proxy artists for shaded regions, and position legends outside the plotting area.

These techniques help make Matplotlib charts clearer, more flexible, and easier to understand.

````python
import matplotlib.pyplot as plt

# Create sample data
months = [1, 2, 3, 4, 5]
sales = [100, 150, 130, 180, 200]
profit = [20, 35, 30, 45, 60]

fig, ax = plt.subplots()

# Plot two lines
ax.plot(months, sales, color="blue", marker="o")
ax.plot(months, profit, color="green", marker="s")

# Retrieve the plotted handles
handles = ax.get_lines()

# Supply custom labels
labels = ["Monthly Sales", "Monthly Profit"]

# Create a custom legend
ax.legend(handles, labels, title="Business Metrics")

ax.set_title("Sales and Profit")
ax.set_xlabel("Month")
ax.set_ylabel("Amount")

plt.show()
````

```python
ax2 = ax1.twinx()
```

The `twinx()` method creates a second Axes object that shares the same x-axis as the primary axes but has an independent y-axis.

In this example:
- `ax1` uses the left y-axis for temperature.
- `ax2` uses the right y-axis for humidity.
- Both axes share the same horizontal time values.


## 9. Retrieving Handles and Labels from Both Axes

```python
handles1, labels1 = ax1.get_legend_handles_labels()
handles2, labels2 = ax2.get_legend_handles_labels()
```

The method `get_legend_handles_labels()` returns two lists:

- `handles`: the plotted objects eligible for the legend.
- `labels`: the corresponding legend text.

The first call retrieves the temperature line and its label from `ax1`.

The second call retrieves the humidity line and its label from `ax2`.

We need both calls because each Axes object maintains its own collection of plotted objects and labels.

## 10. Combining the Legend Entries


The `+` operator concatenates Python lists.

For example, the combined handles list contains the temperature line followed by the humidity line. The combined labels list contains `"Temperature"` followed by `"Humidity"`.

Passing these combined lists to `ax1.legend()` creates a single legend containing entries from both axes.

The legend is created on `ax1`, but it can display handles belonging to both `ax1` and `ax2`.

The argument `loc="upper left"` positions the legend in the upper-left corner of the axes.

# Colormap and Normalization in Matplotlib

## 1. Colormap

A **colormap** defines how numerical values are represented by colors.

For example, in a scatter plot, smaller values might appear purple while larger values appear yellow.

Common colormaps include:
- `viridis`: purple to yellow.
- `plasma`: dark purple to yellow.
- `coolwarm`: blue to red.
- `gray`: shades of gray.

## 2. Normalization

**Normalization** converts data values into a range that a colormap can use, usually between 0 and 1.

For example, if your data ranges from 0 to 100:
- Value 0 maps to 0.
- Value 50 maps to 0.5.
- Value 100 maps to 1.

This allows Matplotlib to map numerical values to colors consistently.

Common normalization methods include:
- `Normalize`: maps values linearly between a minimum and maximum.
- `LogNorm`: useful when positive values span several orders of magnitude.
- `PowerNorm`: applies a power transformation to change how values map to colors.
- `TwoSlopeNorm`: uses a chosen center value, such as zero, to separate two color ranges.

## 3. How Colormap and Normalization Work Together

The two work together to color your data.

1. **Normalization** maps a data value to a normalized value, usually between 0 and 1.
2. **Colormap** converts that normalized value into a color.

For example, with `Normalize(vmin=0, vmax=100)` and the `viridis` colormap, a value of 50 maps to the midpoint of the colormap.

## 4. Key Differences

| Feature | Colormap | Normalization |
|---|---|---|
| Purpose | Selects the colors | Maps data values to the color scale |
| Example | `viridis` | `Normalize(0, 100)` |
| Controls | Which colors are used | How values correspond to colors |

**Remember:** Normalization decides *where* a value falls on the color scale; the colormap decides *which color* it receives.

````python
#Using a colorMap
import matplotlib.pyplot as plt

x = [1, 2, 3, 4, 5]
y = [10, 20, 30, 40, 50]
values = [10, 20, 30, 40, 50]

plt.scatter(x, y, c=values, cmap="viridis")

plt.colorbar(label="Value")
plt.title("Scatter Plot with a Colormap")
plt.xlabel("X")
plt.ylabel("Y")

plt.show()
````

````python
import matplotlib.pyplot as plt
from matplotlib.colors import Normalize

x = [1, 2, 3, 4, 5]
y = [10, 20, 30, 40, 50]
values = [10, 20, 30, 40, 50]

# Map values from 10–50 to the range 0–1
norm = Normalize(vmin=10, vmax=50)

plt.scatter(x, y, c=values, cmap="viridis", norm=norm)

plt.colorbar(label="Value")
plt.title("Scatter Plot with Normalization")
plt.xlabel("X")
plt.ylabel("Y")

plt.show()
````

````python
import matplotlib.pyplot as plt
from matplotlib.colors import LogNorm

x = [1, 2, 3, 4, 5]
y = [10, 20, 30, 40, 50]
values = [1, 10, 100, 1000, 10000]

plt.scatter(
    x, y,
    c=values,
    cmap="plasma",
    norm=LogNorm(vmin=1, vmax=10000)
)

plt.colorbar(label="Value (log scale)")
plt.title("Scatter Plot with Logarithmic Normalization")
plt.xlabel("X")
plt.ylabel("Y")

plt.show()
````

# Mosaic Layout in Matplotlib

## 1. What Is a Mosaic Layout?

A **mosaic layout** is a way to arrange multiple plots in a figure using named areas. Unlike a basic grid, a mosaic layout allows some plots to occupy more space than others.

Matplotlib provides the `subplot_mosaic()` function to create this layout.

## 2. Why Use a Mosaic Layout?

- Arrange multiple plots in one figure.
- Give important plots more space.
- Create layouts with different-sized plots.
- Identify each subplot using a name instead of row and column numbers.

## 3. How Does It Work?

You define a layout using a list of lists. Each letter or name represents a subplot.

For example:

```python id="bzh23q"
layout = [
    ["A", "B"],
    ["C", "C"]
]
```

This creates three subplots:

- `A` occupies the top-left area.
- `B` occupies the top-right area.
- `C` occupies both bottom columns.

The repeated name `"C"` tells Matplotlib that the subplot should span those two positions.

## 4. Important Function

```python id="yzd72p"
fig, axes = plt.subplot_mosaic(layout)
```

- `fig` represents the complete figure.
- `axes` is a dictionary containing the subplot Axes objects.
- The dictionary keys are the names defined in the layout.

For example, `axes["A"]` accesses subplot A, and `axes["C"]` accesses subplot C.

## 5. Key Takeaway

Use `subplot_mosaic()` when you want flexible subplot arrangements with named regions. It is particularly useful for dashboards and reports where plots have different levels of importance.

````python
#Basic mosaic
import matplotlib.pyplot as plt

# Define the layout
layout = [
    ["A", "B"],
    ["C", "C"]
]

# Create the mosaic layout
fig, axes = plt.subplot_mosaic(layout, figsize=(8, 6))

# Plot in subplot A
axes["A"].plot([1, 2, 3], [10, 20, 15])
axes["A"].set_title("Plot A")

# Plot in subplot B
axes["B"].bar(["X", "Y", "Z"], [5, 8, 6])
axes["B"].set_title("Plot B")

# Plot in subplot C, which spans two columns
axes["C"].plot([1, 2, 3, 4], [5, 10, 7, 12])
axes["C"].set_title("Plot C")

fig.tight_layout()
plt.show()
````

# Annotations and Coordinate Systems in Matplotlib

## 1. Annotations

**Annotations** are text labels or notes added to a plot to explain important points or highlight specific information.

For example, you can use an annotation to identify the highest sales value or explain an unusual data point.

### Important function: `ax.annotate()`

```python
ax.annotate("Highest Point", xy=(3, 30), xytext=(2, 35),
            arrowprops=dict(arrowstyle="->"))
```

- `xy`: the point being highlighted.
- `xytext`: the position where the annotation text appears.
- `arrowprops`: controls the appearance of the arrow pointing to the highlighted point.

Annotations can include arrows, colors, and customized text.

## 2. Coordinate Systems

A **coordinate system** determines how positions are specified in a plot.

Matplotlib supports several coordinate systems, depending on what you want to position.

### Common coordinate systems

| Coordinate system | Meaning | Example |
|---|---|---|
| Data coordinates | Positions based on the actual plotted data values. | `(2, 20)` |
| Axes coordinates | Positions relative to the plotting area, where `(0, 0)` is the bottom-left and `(1, 1)` is the top-right. | `(0.5, 0.5)` |
| Figure coordinates | Positions relative to the entire figure, including space outside the axes. | `(0.5, 0.5)` |



## 3. Using Coordinate Systems in Matplotlib

The `transform` parameter tells Matplotlib which coordinate system to use.

For example:

```python
ax.text(0.5, 0.9, "Top Center", transform=ax.transAxes)
```

Here, `ax.transAxes` specifies axes coordinates. Therefore, the text appears near the top-center of the plotting area.

For figure coordinates, use `fig.transFigure`.

## 4. Key Takeaways

- **Annotations** highlight or explain important parts of a plot.
- `ax.annotate()` adds text and can include an arrow.
- **Data coordinates** refer to actual data values.
- **Axes coordinates** position elements relative to the plotting area.
- **Figure coordinates** position elements relative to the complete figure.
- The `transform` parameter selects the coordinate system.

````python
#ADd an annotation with an arrow
import matplotlib.pyplot as plt

fig, ax = plt.subplots()

x = [1, 2, 3, 4, 5]
y = [10, 20, 15, 30, 25]

ax.plot(x, y, marker="o")

# Highlight the highest point
ax.annotate(
    "Highest Point",
    xy=(4, 30),
    xytext=(2.5, 34),
    arrowprops=dict(arrowstyle="->", color="red")
)

ax.set_title("Annotation Example")
ax.set_xlabel("X Values")
ax.set_ylabel("Y Values")

plt.show()
````

# 3D and Specialized Plots in Matplotlib

## 1. What Are 3D Plots?

**3D plots** display data using three dimensions: X, Y, and Z.

They are useful when you want to visualize relationships among three variables.

Examples:
- 3D scatter plots
- 3D line plots
- 3D surface plots

Matplotlib provides 3D plotting capabilities through `mpl_toolkits.mplot3d`.

| Plot type | Purpose |
|---|---|
| 3D scatter plot | Displays individual points in three dimensions. |
| 3D line plot | Shows a line or path through 3D space. |
| 3D surface plot | Displays a continuous surface based on X, Y, and Z values. |
| 3D wireframe plot | Displays a surface using a mesh of lines. |

## 3. What Are Specialized Plots?

**Specialized plots** are chart types designed to visualize particular kinds of data or relationships.

Common examples include:

- **Heatmap:** Uses colors to represent values in a grid.
- **Contour plot:** Shows lines connecting locations with equal values.
- **Polar plot:** Displays data using angles and distances from a center.
- **Stack plot:** Shows how multiple quantities contribute to a total over time.
- **Violin plot:** Shows the distribution and density of numerical data.
- **Hexbin plot:** Groups two-dimensional data into hexagonal bins to show point density.

## 4. Important Functions

| Function | Purpose |
|---|---|
| `ax.scatter()` | Creates a scatter plot, including 3D scatter when used with 3D axes. |

| `ax.stackplot()` | Displays stacked areas. |

## 5. Key Takeaways

- Use **3D plots** when three-dimensional relationships are important.
- Use **specialized plots** when a particular chart type makes your data easier to understand.
- Choose the plot based on your data and the question you want to answer.

````python
#3D scatter plot
import matplotlib.pyplot as plt

# Create a 3D axes
fig = plt.figure()
ax = fig.add_subplot(111, projection="3d")
#So, 111 means 1 row, 1 column, and the first subplot.

# Sample data
x = [1, 2, 3, 4, 5]
y = [5, 3, 6, 2, 4]
z = [10, 15, 12, 20, 18]

# Create the 3D scatter plot
ax.scatter(x, y, z, color="blue", marker="o")

ax.set_xlabel("X")
ax.set_ylabel("Y")
ax.set_zlabel("Z")
ax.set_title("3D Scatter Plot")

plt.show()
````

````python
import matplotlib.pyplot as plt
from matplotlib.animation import FuncAnimation
import numpy as np

# Create a figure and axes
fig, ax = plt.subplots()

# Set the axis limits
ax.set_xlim(0, 10)
ax.set_ylim(0, 10)

# Create an empty point
point, = ax.plot([], [], "ro", markersize=8)

# Function to initialize the animation
def init():
    point.set_data([], [])
    return point,

# Function to update the point in each frame
def update(frame):
    x = frame
    y = frame

    # Update the point's position
    point.set_data([x], [y])

    return point,

# Create the animation
ani = FuncAnimation(
    fig,
    update,
    frames=np.arange(0, 10, 0.1),
    init_func=init,
    interval=50,
    blit=True
)

plt.show()
````
