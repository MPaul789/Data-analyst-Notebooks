# 2. Interactive scatter plot

## Theory
A **scatter plot** represents the relationship between two numerical variables. Each dot represents one observation. Scatter plots are useful for spotting trends, clusters, and unusual values.

An **interactive** scatter plot lets you hover over points to see details, zoom into a region, and pan around the chart. In Plotly, `hover_data` adds extra fields to the tooltip, while `color` can distinguish groups.

## Question
A teacher recorded each student's hours of study, exam score, and class group. Create an interactive scatter plot to explore whether studying longer is associated with higher scores.

````python
# Create a small sample dataset for students.
students = pd.DataFrame({
    "Student": [f"Student {i}" for i in range(1, 16)],
    "Study_Hours": [1.0, 1.5, 2.0, 2.5, 3.0, 3.5, 4.0, 4.5, 5.0, 5.5, 6.0, 6.5, 7.0, 7.5, 8.0],
    "Exam_Score": [42, 48, 51, 55, 59, 63, 65, 70, 73, 75, 81, 84, 86, 91, 94],
    "Group": ["A", "B", "A", "B", "A", "B", "A", "B", "A", "B", "A", "B", "A", "B", "A"]
})

# Build the interactive scatter plot.
fig_scatter = px.scatter(
    students,
    x="Study_Hours",               # horizontal axis
    y="Exam_Score",                # vertical axis
    color="Group",                 # use color to distinguish groups
    hover_name="Student",          # show student name as tooltip heading
    hover_data={"Study_Hours": True, "Exam_Score": True, "Group": True},
    title="Study Hours vs Exam Score",
    labels={"Study_Hours": "Study hours", "Exam_Score": "Exam score"}
)

# Improve the layout and display the chart.
fig_scatter.update_layout(template="plotly_white")
fig_scatter.show()
````

### Code explanation
1. `pd.DataFrame({...})` creates a table with student names, study hours, exam scores, and groups.
2. `px.scatter()` creates one point for each row.
3. `x="Study_Hours"` places study hours on the horizontal axis.
4. `y="Exam_Score"` places exam scores on the vertical axis.
5. `color="Group"` colors the points by class group.
6. `hover_name` and `hover_data` control the information shown when you hover over a point.
7. `update_layout(template="plotly_white")` applies a clean background.
8. `show()` displays the interactive figure.

### Output explanation
The chart should show dots trending upward from left to right. In this sample, students who studied more hours generally earned higher scores. The colored groups help compare the two groups. Hover over a dot to inspect an individual student's values; use the toolbar to zoom, pan, or reset the view.

**Important:** This is illustrative sample data. The pattern in this chart does not prove that study time alone causes higher scores.

# 3. Interactive box plot

## Theory
A **box plot** summarizes a numerical distribution:
- The line inside the box is the median (50th percentile).
- The box spans the first quartile (Q1) to the third quartile (Q3), covering the middle 50% of values.
- The whiskers extend according to the plotting convention; points beyond them may be shown as potential outliers.

Box plots are useful for comparing spread and medians across categories. Plotly makes them interactive, so hovering reveals summary values and individual points when enabled.

## Question
A shop tracks delivery times for three delivery services. Create an interactive box plot to compare their delivery-time distributions and identify possible unusually long deliveries.

````python
# Generate illustrative delivery-time data for three services.
rng = np.random.default_rng(7)
delivery = pd.DataFrame({
    "Service": (["Swift"] * 40) + (["Standard"] * 40) + (["Economy"] * 40),
    "Delivery_Days": np.concatenate([
        rng.normal(1.8, 0.35, 40),   # Swift: shorter typical delivery
        rng.normal(3.2, 0.65, 40),   # Standard
        rng.normal(5.0, 1.0, 40)    # Economy: longer and more variable
    ])
})

# Delivery days cannot be negative, so replace any negative generated values with a small positive value.
delivery["Delivery_Days"] = delivery["Delivery_Days"].clip(lower=0.2)

# Draw a box plot and show the underlying observations as points.
fig_box = px.box(
    delivery,
    x="Service",
    y="Delivery_Days",
    color="Service",
    points="outliers",  # show points outside the whiskers
    title="Delivery Time by Service",
    labels={"Service": "Delivery service", "Delivery_Days": "Delivery time (days)"}
)
fig_box.update_layout(template="plotly_white", showlegend=False)
fig_box.show()
````

### Code explanation
- `np.random.default_rng(7)` creates a random-number generator with a fixed seed.
- `np.concatenate()` joins the generated delivery-time samples into one array.
- The `Service` column labels which service each observation belongs to.
- `.clip(lower=0.2)` prevents impossible negative delivery times in this illustrative dataset.
- `px.box()` creates a box for each service.
- `points="outliers"` displays observations classified as outliers by the box-plot rule.
- `showlegend=False` hides a redundant legend because the service is already labeled on the x-axis.

### Output explanation
You should see three boxes. Swift is designed to have the shortest typical delivery time, while Economy has a longer and wider distribution. Wider boxes and longer whiskers indicate greater variability. Dots beyond the whiskers are potential outliers, not necessarily data errors.

# 4. Interactive box plot and histogram together

## Theory
A **histogram** groups numerical values into bins and shows how many observations fall in each bin. It helps reveal the shape of a distribution, including concentration, skewness, and multiple peaks.

A box plot gives a compact summary, while a histogram shows more detail about frequencies. Putting them side by side lets us compare both views of the same data. Plotly Graph Objects supports subplots, and both charts remain interactive.

## Question
A company records employee commute times. Create a figure with a histogram and a box plot so that you can inspect the distribution and spot possible long commutes.

````python
from plotly.subplots import make_subplots

# Create illustrative commute-time data in minutes.
rng = np.random.default_rng(21)
commute_minutes = np.concatenate([
    rng.normal(25, 7, 85),     # most employees have shorter commutes
    rng.normal(55, 10, 15)     # a smaller group has longer commutes
])
commute_minutes = np.clip(commute_minutes, 5, None)

# Create two coordinated panels: histogram on the left, box plot on the right.
fig_combo = make_subplots(
    rows=1, cols=2,
    subplot_titles=("Commute Time Histogram", "Commute Time Box Plot"),
    column_widths=[0.68, 0.32],
    horizontal_spacing=0.12
)

# Add the histogram, which counts observations in time bins.
fig_combo.add_trace(
    go.Histogram(
        x=commute_minutes,
        nbinsx=15,
        name="Employees",
        marker_line_width=1,
        hovertemplate="Commute range: %{x}<br>Count: %{y}<extra></extra>"
    ),
    row=1, col=1
)

# Add a horizontal box plot to summarize the same observations.
fig_combo.add_trace(
    go.Box(
        x=commute_minutes,
        name="Commute times",
        boxpoints="outliers",
        orientation="h",
        hovertemplate="Value: %{x:.1f} minutes<extra></extra>"
    ),
    row=1, col=2
)

# Add titles and clean layout.
fig_combo.update_xaxes(title_text="Commute time (minutes)", row=1, col=1)
fig_combo.update_yaxes(title_text="Number of employees", row=1, col=1)
fig_combo.update_xaxes(title_text="Minutes", row=1, col=2)
fig_combo.update_layout(
    title_text="Employee Commute-Time Distribution",
    template="plotly_white",
    showlegend=False,
    height=480
)
fig_combo.show()
````

### Code explanation
1. `make_subplots()` creates a one-row, two-column figure.
2. `go.Histogram()` groups commute times into bins; `nbinsx=15` requests about 15 bins.
3. `hovertemplate` customizes the tooltip. `%{x}` and `%{y}` are values supplied by Plotly.
4. `go.Box()` summarizes the same commute-time values in a horizontal box plot.
5. `boxpoints="outliers"` asks Plotly to display points beyond the whiskers.
6. `update_xaxes()` and `update_yaxes()` label each panel's axes.
7. `height=480` makes the combined figure tall enough to read.

### Output explanation
The histogram shows how many employees fall into each commute-time interval. The box plot shows the median, middle 50%, spread, and possible outliers. In this sample, most commutes cluster around a shorter duration, while a smaller group has longer commutes. Compare the histogram's shape with the box plot's whiskers and outlier points.

# 5. Interactive line chart with a range slider

## Theory
A **line chart** connects observations in order and is especially useful for time-series data, such as daily sales, temperature, or website visits.

A **range slider** appears below the x-axis and lets you select a smaller time window without rebuilding the chart. A **range selector** provides shortcut buttons such as 1 month, 3 months, or all data. These controls are useful when a time series contains many observations.

## Question
A store tracks daily sales for one year. Create an interactive line chart with a range slider and quick-selection buttons so a user can inspect a specific period.

````python
# Generate one year of illustrative daily sales data.
dates = pd.date_range(start="2025-01-01", periods=365, freq="D")
rng = np.random.default_rng(100)

# Add a gentle trend, weekly variation, and random noise.
trend = np.linspace(100, 150, len(dates))
weekly_pattern = 12 * np.sin(2 * np.pi * np.arange(len(dates)) / 7)
noise = rng.normal(0, 8, len(dates))
sales = trend + weekly_pattern + noise

sales_df = pd.DataFrame({
    "Date": dates,
    "Sales": sales.round(2)
})

# Create the interactive line chart.
fig_line = go.Figure()

fig_line.add_trace(
    go.Scatter(
        x=sales_df["Date"],
        y=sales_df["Sales"],
        mode="lines",
        name="Daily sales",
        line=dict(width=2),
        hovertemplate="Date: %{x|%d %b %Y}<br>Sales: %{y:.2f}<extra></extra>"
    )
)

# Add a range slider and preset date-range buttons.
fig_line.update_layout(
    title="Daily Sales Over One Year",
    template="plotly_white",
    xaxis=dict(
        title="Date",
        rangeslider=dict(visible=True),
        rangeselector=dict(
            buttons=[
                dict(count=1, label="1m", step="month", stepmode="backward"),
                dict(count=3, label="3m", step="month", stepmode="backward"),
                dict(count=6, label="6m", step="month", stepmode="backward"),
                dict(step="all", label="All")
            ]
        )
    ),
    yaxis_title="Sales (illustrative units)"
)

fig_line.show()
````

### Code explanation
- `pd.date_range()` creates 365 consecutive dates starting on January 1, 2025.
- `np.linspace()` creates a gradual upward trend.
- `np.sin()` adds a repeating weekly pattern.
- `rng.normal()` adds realistic-looking random variation.
- `go.Scatter(..., mode="lines")` draws the time series as a line.
- `rangeslider=dict(visible=True)` displays the draggable range slider.
- `rangeselector` creates quick buttons for the previous 1, 3, or 6 months and the full year.
- `hovertemplate` formats the date and sales value shown on hover.

### Output explanation
The chart displays daily sales across the year. Drag the handles on the range slider to zoom into a date range, or select `1m`, `3m`, `6m`, or `All`. Hover over the line to read the date and sales value. The sample has a mild upward trend, weekly fluctuations, and random noise; it is generated data, not real store records.
