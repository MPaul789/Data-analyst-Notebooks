# Formulas for Descriptive Statistics

## 1. Mean

The mean is the average of all observations.

$$
\bar{x} = \frac{\sum_{i=1}^{n} x_i}{n}
$$

Where:

- $\bar{x}$ = Mean
- $x_i$ = Individual observation
- $n$ = Number of observations
- $\sum$ = Sum of all observations


---

## 2. Median

The median is the middle value after arranging the data in ascending order.

### If $n$ is odd:

$$
\text{Median} = x_{\frac{n+1}{2}}
$$

### If $n$ is even:

$$
\text{Median} =
\frac{
x_{\frac{n}{2}} + x_{\frac{n}{2}+1}
}{2}
$$


---

## 3. Mode

The mode is the value that occurs most frequently.

$$
\text{Mode} = \text{Most Frequently Occurring Value}
$$


# When Should We Use Mean, Median, or Mode?

Choosing between **Mean, Median, and Mode** depends on the type and distribution of the data.

---

## 1. When to Use Mean

Use the **Mean** when:

- The data is **numerical**.
- The data is **fairly symmetric**.
- There are **no extreme outliers**.
- You want to consider **every value** in the dataset.

### Example

Consider the marks:

$$
60,\ 62,\ 65,\ 68,\ 70
$$

The data is fairly balanced, so the **Mean** is a good measure of central tendency.

> **Rule:** Balanced numerical data → **Mean**

---

## 2. When to Use Median

Use the **Median** when:

- The data is **numerical**.
- The data contains **outliers**.
- The data is **skewed**.
- You want a value that is **less affected by extreme values**.

### Example

Consider the salaries:

$$
25,000,\ 28,000,\ 30,000,\ 32,000,\ 500,000
$$

The value ₹500,000 is an **outlier**.

It will pull the Mean upward, while the Median will remain more representative of a typical salary.

> **Rule:** Skewed data or outliers → **Median**

---

## 3. When to Use Mode

Use the **Mode** when:

- You want to find the **most frequently occurring value**.
- The data is **categorical**.
- Mean and Median do not make meaningful sense.
- You want to identify the **most popular choice**.

### Example 1: Numerical Data

Number of products purchased:

$$
1,\ 2,\ 2,\ 2,\ 3,\ 4,\ 5
$$

The value **2** occurs most frequently.

$$
\text{Mode} = 2
$$

### Example 2: Categorical Data

City preferences:

$$
Delhi,\ Mumbai,\ Delhi,\ Pune,\ Delhi,\ Mumbai
$$

The most frequently occurring city is **Delhi**.

$$
\text{Mode} = \text{Delhi}
$$

> **Rule:** Most common value or category → **Mode**

---

# Quick Decision Rule

| Data Situation | Best Choice |
|---|---|
| Numerical + fairly symmetric + no outliers | **Mean** |
| Numerical + skewed data | **Median** |
| Numerical + extreme outliers | **Median** |
| Categorical data | **Mode** |
| Most frequently occurring value | **Mode** |

### Easy Way to Remember

$$
\boxed{\text{Balanced Data} \rightarrow \text{Mean}}
$$

$$
\boxed{\text{Skewed / Outliers} \rightarrow \text{Median}}
$$

$$
\boxed{\text{Most Common} \rightarrow \text{Mode}}
$$

# Probability Distribution

A **probability distribution** describes all the possible outcomes of a random variable and the probability associated with each outcome.

In simple words, it tells us **what outcomes can occur and how likely each outcome is**.

### Example

Suppose we toss a fair coin.

Possible outcomes:

- Heads
- Tails

The probability of each outcome is:

$$
P(\text{Heads}) = \frac{1}{2} = 0.5
$$

$$
P(\text{Tails}) = \frac{1}{2} = 0.5
$$

The sum of all possible outcome probabilities is 1.

$$
\sum P(X=x) = 1
$$

```python
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
from scipy import stats

# Set overall plotting style
sns.set_theme(style="whitegrid")

# ==========================================
# 1. Binomial Distribution
# ==========================================
# Fixed trials, two outcomes per trial (success/failure); e.g., coin tosses, pass/fail[cite: 4]
n = 10
p = 0.5
k = np.arange(0, n + 1)
binomial_pmf = stats.binom.pmf(k, n, p)

# Specific example from the image[cite: 4]
print("Binomial PMF (k=3, n=10, p=0.5):", stats.binom.pmf(3, 10, 0.5))

plt.figure(figsize=(8, 4))
sns.barplot(x=k, y=binomial_pmf, color='skyblue')
plt.title('Binomial Distribution (n=10, p=0.5)')
plt.xlabel('Number of Successes (k)')
plt.ylabel('Probability (PMF)')
plt.show()



```



## Conditions for Binomial Distribution

A binomial experiment must satisfy these four conditions:

1. **Fixed number of trials:** The number of trials is decided in advance.
2. **Two possible outcomes:** Each trial has two outcomes, such as success or failure.
3. **Independent trials:** The outcome of one trial does not affect another.
4. **Constant probability:** The probability of success remains the same in every trial.

## Formula

The probability of getting exactly $k$ successes in $n$ trials is:

$$
P(X=k) = \binom{n}{k}p^k(1-p)^{n-k}
$$

Where:

- $P(X=k)$ = Probability of exactly $k$ successes
- $n$ = Total number of trials
- $k$ = Number of successes
- $p$ = Probability of success in one trial
- $1-p$ = Probability of failure
- $\binom{n}{k}$ = Number of ways to choose $k$ successes from $n$ trials

## Understanding the Python Code

### 1. Import Libraries

- `numpy` is used to create numerical values.
- `matplotlib.pyplot` is used to display plots.

### 2. Set the Plotting Style

`sns.set_theme(style="whitegrid")` applies a white background with grid lines to make the chart easier to read.

### 3. Define the Parameters

- `n = 10`: There are 10 trials.
- `p = 0.5`: The probability of success in each trial is 0.5.
- `k = np.arange(0, n + 1)`: Creates possible success counts from 0 to 10.

Why 0 to 10? Because in 10 trials, we can get anywhere from zero successes to 10 successes.

### 4. Calculate Binomial Probabilities

`stats.binom.pmf(k, n, p)` calculates the probability of getting exactly each possible number of successes.

PMF stands for **Probability Mass Function**.

For example:

`stats.binom.pmf(3, 10, 0.5)` calculates the probability of getting exactly 3 successes in 10 independent trials, with a success probability of 0.5 per trial.

The result is approximately:

$$
P(X=3) = 0.1172
$$

Therefore, the probability is approximately **11.72%**.

# 3. Normal Distribution

## What is a Normal Distribution?

A **Normal Distribution** is a continuous probability distribution in which the data is distributed symmetrically around the mean, forming a bell-shaped curve.

In simple words, most values are close to the average, while fewer values occur far away from the average.

### Example

Suppose the marks of 1,000 students follow an approximately normal distribution.

- Most students score close to the average.
- A smaller number of students score very low.
- A smaller number of students score very high.

The graph looks like a bell, with the highest point near the mean.

## Characteristics of a Normal Distribution

1. **Bell-shaped:** The curve resembles a bell.
2. **Symmetrical:** The left and right sides are approximately mirror images.
3. **Mean = Median = Mode:** All three measures are equal at the center.
4. **Total area under the curve = 1:** This represents 100% probability.
5. **Tails extend indefinitely:** The curve approaches the horizontal axis but never touches it.
6. **Defined by two parameters:**
   - Mean ($\mu$): Determines the center of the distribution.
   - Standard deviation ($\sigma$): Determines how spread out the distribution is.

## The 68–95–99.7 Rule

For a normal distribution:

- Approximately **68%** of observations lie within 1 standard deviation of the mean.
- Approximately **95%** lie within 2 standard deviations.
- Approximately **99.7%** lie within 3 standard deviations.

For a distribution with mean $\mu$ and standard deviation $\sigma$:

$$
P(\mu-\sigma \leq X \leq \mu+\sigma) \approx 0.68
$$

$$
P(\mu-2\sigma \leq X \leq \mu+2\sigma) \approx 0.95
$$

$$
P(\mu-3\sigma \leq X \leq \mu+3\sigma) \approx 0.997
$$

## Example Using Student Marks

Suppose:

- Mean marks = 70
- Standard deviation = 10

Then:

| Interval | Approximate Percentage | Marks |
|---|---:|---|
| Mean ± 1 SD | 68% | 60 to 80 |
| Mean ± 2 SD | 95% | 50 to 90 |
| Mean ± 3 SD | 99.7% | 40 to 100 |

This means approximately 68% of students score between 60 and 80, assuming their marks follow a normal distribution.

## Normal Distribution Formula

The probability density function (PDF) is:

$$
f(x)=\frac{1}{\sigma\sqrt{2\pi}}
e^{-\frac{(x-\mu)^2}{2\sigma^2}}
$$

Where:

- $x$ = Observation
- $\mu$ = Mean
- $\sigma$ = Standard deviation
- $e$ = Euler's number
- $\pi$ = Mathematical constant pi

You do not need to memorize this formula for basic data analysis. Focus on understanding the shape, mean, standard deviation, and probability intervals.

# Central Limit Theorem (CLT) and Sampling Distribution

## 1. Population and Sample

**Population:** The complete group of people or objects we want to study.

**Sample:** A smaller group selected from the population.

**Sample Mean:** The average value calculated from a sample.

### Example

Suppose a college has 10,000 students, and we want to calculate their average marks.

- Population: Marks of all 10,000 students.
- Sample: Marks of 30 randomly selected students.
- Sample Mean: Average marks of those 30 students.

Different samples may produce different sample means.

For example:
- Sample 1: Average = 65
- Sample 2: Average = 68
- Sample 3: Average = 66

This variation is natural because each sample may contain different students.

---

## 2. Sampling Distribution of Sample Means

A **sampling distribution of sample means** is the distribution formed by the means of many samples of the same size taken from a population.

### How does it work?

1. Randomly select 30 students.
2. Calculate their average marks.
3. Select another random sample of 30 students.
4. Calculate its average marks.
5. Repeat this process many times.
6. Plot a histogram of all the calculated sample means.

The resulting histogram represents the sampling distribution of sample means.

**Important:** We are plotting the sample averages, not the individual students' marks.

### Why is it useful?

A sampling distribution helps us understand:

- How much sample means vary.
- How reliable a sample-based estimate is.
- How standard error is calculated.
- How confidence intervals and hypothesis tests work.

---

## 3. Central Limit Theorem (CLT)

The **Central Limit Theorem** states that when we repeatedly take sufficiently large random samples from a population and calculate their means, the distribution of those sample means tends to become approximately normal, even if the original population is not normally distributed.

### Example

Suppose the marks of students in a college are right-skewed.

This means many students may have moderate marks, while a few students have very high marks.

If we repeatedly select random samples of 30 students and calculate their average marks, the distribution of these sample averages will tend to become approximately bell-shaped.

### Important Points

- The original population may be skewed.
- The distribution of sample means tends to become normal as the sample size increases.
- The sample means tend to cluster around the population mean.
- Larger samples generally produce less variation in sample means.

**Remember:** CLT does not say that the original data becomes normally distributed. It describes the distribution of sample means.

---

## 4. Mean of the Sampling Distribution

### Formula

Mean of sample means = Population mean

Mathematically:

μₓ̄ = μ

Where:

- μₓ̄ = Mean of the sampling distribution of sample means.
- μ = Population mean.
- x̄ = Mean of an individual sample.

### Example

Suppose the population mean marks are 70.

If we repeatedly take random samples and calculate their means, the average of those sample means will tend towards 70.

This means the sample mean is an unbiased estimator of the population mean under the usual random-sampling conditions.

---

## 5. Standard Error

**Standard error** measures how much sample means typically vary from one another.

### Formula

SE = σ / √n

Where:

- SE = Standard error of the sample mean.
- σ = Population standard deviation.
- n = Sample size.

### Example

Suppose:

- Population standard deviation = 12
- Sample size = 36

Calculation:

SE = 12 / √36

SE = 12 / 6

SE = 2

Therefore, the standard error is 2 marks.

### How does sample size affect standard error?

| Sample Size | Standard Error |
|---:|---:|
| 4 | 12 / √4 = 6 |
| 16 | 12 / √16 = 3 |
| 36 | 12 / √36 = 2 |
| 144 | 12 / √144 = 1 |

**Conclusion:** As sample size increases, standard error decreases. Therefore, sample means generally become more concentrated around the population mean.

Note: This formula assumes independent observations and uses the population standard deviation. When the population standard deviation is unknown, we commonly estimate standard error using the sample standard deviation.

---

## 6. Rule of Thumb: Sample Size of 30

A common rule of thumb is that a sample size of 30 or more may be sufficient for the sampling distribution of the mean to become approximately normal.

However, this is not a universal rule.

- If the population is normally distributed, the sample mean is normally distributed for any sample size, assuming independent random sampling.
- If the population is moderately skewed, a sample size around 30 may be sufficient.
- If the population is highly skewed or contains extreme outliers, a larger sample may be needed.

The required sample size depends on the population distribution and sampling conditions.

---

## 7. Difference Between Population Distribution and Sampling Distribution

| Population Distribution | Sampling Distribution |
|---|---|
| Represents individual population values. | Represents a statistic, such as sample means. |
| May be normal, skewed, or irregular. | The distribution of sample means tends towards normality under CLT conditions. |
| Describes variation among individuals. | Describes variation among sample means. |
| Example: Marks of all students. | Example: Average marks from thousands of samples. |

---

## 8. Applications of CLT in Data Analytics

The Central Limit Theorem is useful in:

1. **Confidence Intervals:** Estimating a range of plausible values for a population mean.
2. **Hypothesis Testing:** Evaluating evidence against a statistical claim.
3. **Standard Error:** Measuring the variability of sample-based estimates.
4. **Quality Control:** Monitoring production measurements using samples.
5. **Business Analytics:** Estimating average customer spending or delivery time from samples.

---

## 9. Summary

- A population contains all the values we want to study.
- A sample is a subset of the population.
- A sample mean is the average calculated from a sample.
- A sampling distribution is formed from the means of many samples.
- The Central Limit Theorem explains why the distribution of sample means tends towards a normal distribution.
- The mean of the sampling distribution equals the population mean under standard random-sampling conditions.
- Standard error measures the variability of sample means.
- Increasing sample size generally reduces standard error and improves the precision of the sample mean.

# Confidence Interval (CI)

## 1. What Is a Confidence Interval?

A **Confidence Interval (CI)** is a range of values used to estimate an unknown population parameter, such as the population mean.

When we collect a sample from a population, the sample mean may not be exactly equal to the population mean. Therefore, instead of relying on a single estimate, we calculate a range that accounts for sampling uncertainty.

### Example

Suppose we want to estimate the average marks of all students in a college.

- Population: All students in the college.
- Sample: 100 randomly selected students.
- Sample mean: 70 marks.
- 95% Confidence Interval: 68.04 to 71.96 marks.

The interval gives us a range of plausible values for the true population mean based on our sample and the assumptions of our method.

---

## 2. Why Do We Need a Confidence Interval?

A confidence interval helps us:

- Estimate an unknown population mean.
- Understand the uncertainty associated with a sample estimate.
- Measure the precision of an estimate.
- Make informed decisions using sample data.
- Support statistical analysis and hypothesis testing.

**Important:** A narrower confidence interval generally indicates a more precise estimate, while a wider interval indicates greater uncertainty.

---

## 3. What Does a 95% Confidence Level Mean?

A 95% confidence level means that if we repeatedly collected random samples and calculated a confidence interval using the same method, approximately 95% of those intervals would contain the true population mean in the long run.

For example, if we calculated 100 confidence intervals from 100 different random samples, approximately 95 intervals would be expected to contain the true population mean.

**Remember:** The population mean is fixed. The confidence interval changes from sample to sample. The 95% refers to the long-run success rate of the method, not a 95% probability that the fixed population mean lies inside one already-calculated interval.

---

## 4. Confidence Interval Formula

The basic formula is:

**Confidence Interval = Estimate ± Margin of Error**

This means we start with our sample estimate and calculate how far to extend the interval in both directions.

- Lower Limit = Estimate − Margin of Error
- Upper Limit = Estimate + Margin of Error

For estimating a population mean, the sample mean is usually our estimate.

### Formula for a Z-Confidence Interval

When the population standard deviation is known:

$$
\boxed{\text{CI} = \bar{x} \pm z_{\alpha/2}\frac{\sigma}{\sqrt{n}}}
$$
This formula may look complicated, but it contains only three main parts:

**Confidence Interval = Sample Mean ± (Critical Value × Standard Error)**

In other words:

1. Calculate the sample mean.
2. Calculate the standard error.
3. Multiply the standard error by the critical value to get the margin of error.
4. Subtract and add the margin of error to the sample mean.



### Understanding the Symbols in the Confidence Interval Formula

| Symbol | Meaning |
|---|---|
| $\bar{x}$ | Sample mean — average calculated from the sample |
| $z_{\alpha/2}$ | Critical Z-value determined by the confidence level |
| $\sigma$ | Population standard deviation |
| $n$ | Sample size — number of observations |
| $\sqrt{n}$ | Square root of the sample size |
| $\frac{\sigma}{\sqrt{n}}$ | Standard error of the sample mean |
| $z_{\alpha/2}\times\frac{\sigma}{\sqrt{n}}$ | Margin of error |

For a **95% confidence level**, the critical Z-value is approximately **1.96**.

Therefore, the formula becomes:

$$
\boxed{\text{CI}=\bar{x}\pm1.96\times\frac{\sigma}{\sqrt{n}}}
$$
For a **95% confidence level**, the critical z-value is approximately **1.96**.



## 5. Understanding Standard Error

The standard error measures how much sample means typically vary from one another across repeated samples.

### Formula

$$
\boxed{SE = \frac{\sigma}{\sqrt{n}}}
$$

In simple words:

**Standard Error = Population Standard Deviation ÷ Square Root of Sample Size**

### Example

Suppose:

- Population standard deviation = 10
- Sample size = 100

Step 1: Calculate the square root of the sample size.

$$
\[
\sqrt{100}=10
\]
$$
Step 2: Divide the standard deviation by the square root of the sample size.

$$
\[
SE=\frac{10}{10}=1
\]
$$ 
Therefore, the standard error is **1**.

A larger sample size generally produces a smaller standard error, which helps us estimate the population mean more precisely.

---

## 6. Understanding the Margin of Error

The margin of error tells us how far the confidence interval extends above and below the sample mean.

### Formula

$$
\boxed{\text{Margin of Error}=\text{Critical Value}\times SE}
$$

For a 95% Z-confidence interval:

$$
ME=1.96\times SE
$$

If the standard error is 1:

$$
ME=1.96\times1=1.96
$$

This means we extend the interval by 1.96 on either side of the sample mean.

---

## 7. Step-by-Step Example: Calculate a 95% Confidence Interval

Suppose we want to estimate the average marks of all students in a college.

We collect a random sample and obtain the following information:

- Sample mean = 70 marks
- Population standard deviation = 10 marks
- Sample size = 100 students
- Confidence level = 95%
- Critical z-value = 1.96

### Step 1: Calculate the Standard Error

Formula:

$$
\[
SE=\frac{\sigma}{\sqrt{n}}
\]
$$

Substitute the values:

$$
\[
SE=\frac{10}{\sqrt{100}}
\]
$$

$$
\[
SE=\frac{10}{10}=1
\]
$$

**Standard Error = 1 mark**

### Step 2: Calculate the Margin of Error

Formula:

$$
\[
ME=z\times SE
\]
$$

Substitute the values:

$$
\[
ME=1.96\times1=1.96
\]
$$

**Margin of Error = 1.96 marks**

### Step 3: Calculate the Lower Limit

Formula:

$$
\[
\text{Lower Limit}=\bar{x}-ME
\]
$$

Substitute the values:

$$
\[
70-1.96=68.04
\]
$$

**Lower Limit = 68.04 marks**

### Step 4: Calculate the Upper Limit

Formula:

$$
\[
\text{Upper Limit}=\bar{x}+ME
\]
$$

Substitute the values:

$$
\[
70+1.96=71.96
\]
$$

**Upper Limit = 71.96 marks**

### Final Answer

$$
\[
\boxed{68.04\leq\mu\leq71.96}
\]
$$

The 95% confidence interval for the population mean is **68.04 to 71.96 marks**.

### Interpretation

Based on our sample, we estimate that the true average marks of all students lies between 68.04 and 71.96 marks, using a method that captures the true population mean in approximately 95% of repeated samples under the stated assumptions.

## 8. What Factors Affect a Confidence Interval?

### A. Sample Size

A larger sample size generally reduces the standard error and produces a narrower confidence interval.

For example, increasing the sample size from 25 to 100 reduces the standard error by half, assuming the population standard deviation remains the same.

### B. Confidence Level

A higher confidence level produces a wider interval because the method must capture the true population parameter more often in repeated sampling.

| Confidence Level | Approximate Critical Z-Value |
|---|---:|
| 90% | 1.645 |
| 95% | 1.96 |
| 99% | 2.576 |

For the same data, a 99% confidence interval is wider than a 95% confidence interval, which is wider than a 90% confidence interval.

### C. Standard Deviation

A larger population standard deviation indicates greater variability in the data and generally produces a wider confidence interval.

## 10. Relationship Between Confidence Interval and Central Limit Theorem

The **Central Limit Theorem (CLT)** states that, under suitable conditions, the distribution of sample means tends towards a normal distribution as the sample size increases.

This helps us construct confidence intervals because it allows us to describe the sampling uncertainty around a sample mean using a normal approximation when appropriate.

Confidence intervals use the sample mean, standard error, and a critical value to estimate a range for the population mean.
