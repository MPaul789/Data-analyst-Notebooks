# Python Data Types

A **data type** tells Python what kind of value a variable contains.

### 🔹 1. Numeric Types

| Data Type | Description | Example |
|---|---|---|
| `int` | Whole numbers | `10`, `-5`, `100` |
| `float` | Decimal numbers | `10.5`, `3.14` |
| `complex` | Complex numbers | `2 + 3j` |

```python
age = 25
price = 99.99
z = 2 + 3j


## 2. `float` — Floating Point Number

Used to store **decimal numbers**.

````python
price = 99.99
percentage = 85.5
temperature = 36.6

print(price)
print(percentage)
print(temperature)
````

## 3. `complex` — Complex Number

Used to store **complex numbers** with a real part and an imaginary part.

````python
z = 2 + 3j

print(z)
print(z.real)
print(z.imag)
````

## 4. `bool` — Boolean

Used to store **True or False** values.

```python
is_logged_in = True
is_admin = False

print(is_logged_in)
print(is_admin)

## 5. `str` — String

Used to store **text or a sequence of characters**.

```python
name = "Rahul"
message = "Welcome to Python"

print(name)
print(message)

## 6. `list` — List

Used to store **multiple values in a single variable**.

A list is **ordered and mutable**, and it can contain duplicate values.

```python
marks = [80, 90, 75, 80]

print(marks)
print(marks[0])

## 7. `tuple` — Tuple

Used to store **multiple values in a single variable**.

A tuple is **ordered and immutable**, which means its values cannot be changed after creation.

```python
coordinates = (10, 20, 30)

print(coordinates)
print(coordinates[0])

## 8. `range` — Range

Used to generate a **sequence of numbers**.

It is commonly used with **loops**.



````python
numbers = range(1, 6)

print(list(numbers))
````

````python
## 9. `set` — Set

Used to store **multiple unique values**.

A set is **unordered and mutable**, and duplicate values are automatically removed.

````

````python
numbers = {10, 20, 30, 10, 20}

print(numbers)
````

## 10. `frozenset` — Frozen Set

Used to store **unique values** in an **immutable set**.

Unlike a normal `set`, a `frozenset` cannot be modified after creation.



````python
numbers = frozenset({10, 20, 30})

print(numbers)
````

## 11. `dict` — Dictionary

Used to store data in **key-value pairs**.



````python

student = {
    "name": "Rahul",
    "age": 22,
    "marks": 85
}

print(student)
print(student["name"])
````

##  12. `NoneType` — None

Used to represent the **absence of a value**.


````python
result = None

print(result)
print(type(result))
````

## 13. `bytes` — Bytes

Used to store **immutable binary data**.
### 📌 Important Points

- `bytes` is used to store **binary data**.
- A `bytes` object is **immutable**, so its values cannot be changed.
- Each element in a `bytes` object is an **integer from 0 to 255**.
- It is commonly used when working with **files, images, network communication, and encoded data**.
- Bytes literals are created using the **`b` prefix**.



````python

data = b"Hello"

print(data)
print(data[0])
````

## 14. `bytearray` — Byte Array

Used to store **binary data that can be modified**.

Unlike `bytes`, a `bytearray` is **mutable**.



````python

data = bytearray(b"Hello")

print(data)

data[0] = 74

print(data)

#74 is the ASCII value of J.So H is replaced with J
````

### 📌 Important Points

- `memoryview` provides a **view of an existing binary object's memory**.
- It does **not create a separate copy** of the original data.
- It can be created from objects that support the **buffer protocol**, such as `bytes` and `bytearray`.
- It is useful for working with **large amounts of binary data efficiently**.
- Individual elements can be accessed using **indexing**.
- Each element is returned as an **integer between 0 and 255**.
- A `memoryview` can be **sliced** to access a portion of the data.
- With a mutable object such as `bytearray`, changes made through the `memoryview` can affect the original data.
- `memoryview` is commonly useful in **file handling, networking, and binary data processing**.




````python
data = bytearray(b"Hello")

view = memoryview(data)

print(view[0])

view[0] = 74

print(data)
````

Here, memoryview accesses the same memory as data. Changing view[0] also changes the original bytearray

**LOOPING**

##  1. `for` Loop

A `for` loop is used to **iterate over a sequence or collection**.

```python
numbers = [10, 20, 30, 40, 50]

for num in numbers:
    print(num)
```

**Output:**

```text
10
20
30
40
50
```

### 📌 Key Point

The loop takes **one element at a time** from the collection and executes the loop body.

## 2. `for` Loop with `range()`

`range()` is commonly used when we want to repeat a block of code a specific number of times.

```python
for i in range(1, 6):
    print(i)
```

**Output:**

```text
1
2
3
4
5
```

### 📌 Key Point

`range(start, stop)` includes the `start` value but **does not include the `stop` value**.

##  3. `while` Loop

A `while` loop repeatedly executes a block of code **as long as a condition is `True`**.

```python
count = 1

while count <= 5:
    print(count)
    count += 1
```

**Output:**

```text
1
2
3
4
5
```

### 📌 Key Point

The condition is checked before every iteration.

> ⚠️ Make sure the condition eventually becomes `False`, otherwise you may create an **infinite loop**.

##  5. `break` Statement

The `break` statement is used to **immediately stop a loop**.

```python
for num in range(1, 10):
    if num == 5:
        break

    print(num)
```

**Output:**

```text
1
2
3
4
```

### 📌 Key Point

When `num` becomes `5`, `break` terminates the loop completely.

## 6. `continue` Statement

The `continue` statement **skips the current iteration** and moves to the next iteration.

```python
for num in range(1, 6):
    if num == 3:
        continue

    print(num)
```

**Output:**

```text
1
2
4
5
```

### 📌 Key Point

`continue` does **not stop the loop**. It only skips the current iteration.

##  7. `pass` Statement

`pass` is used when we want to **do nothing** for the moment.

```python
for num in range(1, 6):
    if num == 3:
        pass

    print(num)
```

**Output:**

```text
1
2
3
4
5
```

### 📌 Key Point

Unlike `continue`, `pass` **does not skip the iteration**.

It simply acts as a placeholder where Python expects a statement.

## 8. `else` with a Loop

Python allows an `else` block to be used with `for` and `while` loops.

The `else` block executes when the loop **finishes normally**.

```python
for num in range(1, 4):
    print(num)
else:
    print("Loop completed")
```

**Output:**

```text
1
2
3
Loop completed
```

### 📌 Important

If the loop is terminated using `break`, the `else` block **will not execute**.

```python
for num in range(1, 4):
    if num == 2:
        break

    print(num)
else:
    print("Loop completed")
```

**Output:**

```text
1
```

## 9. `enumerate()` with a Loop

`enumerate()` allows us to get both the **index and the value** while looping.

```python
students = ["Rahul", "Priya", "Amit"]

for index, name in enumerate(students):
    print(index, name)
```

**Output:**

```text
0 Rahul
1 Priya
2 Amit
```

### 📌 Key Point

Instead of manually maintaining a counter, `enumerate()` gives us the **index and value together**.

## 10. `zip()` with a Loop

`zip()` allows us to iterate over **multiple sequences at the same time**.

```python
names = ["Rahul", "Priya", "Amit"]
marks = [85, 92, 78]

for name, mark in zip(names, marks):
    print(name, mark)
```

**Output:**

```text
Rahul 85
Priya 92
Amit 78
```

### 📌 Key Point

`zip()` combines corresponding elements from multiple sequences.
