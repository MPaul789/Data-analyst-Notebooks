### 📌 Dictionary preserves insertion order

> **Ordered does not mean sorted.**

```python
data = {
    "z": 10,
    "a": 20,
    "m": 30
}

Python maintains the insertion order:

z → a → m

It does not automatically sort the keys alphabetically.

So:
- "z" remains first
- "a" remains second
- "m" remains third
💡 Remember:
Insertion order ≠ Alphabetical order

**FILE HANDLING**
Common file modes:
"r"  -> Read
"w"  -> Write (creates a new file / overwrites existing content)
"a"  -> Append (adds content at the end of the file)

1. CREATE / WRITE TO A FILE
"w" mode is used to write data into a file.
If the file does not exist, Python creates it.
If it already exists, its previous content is overwritten.

````python
file = open("students.txt", "w")

file.write("Rahul,85\n")
file.write("Priya,92\n")
file.write("Amit,78\n")

file.close()

print("File created and data written successfully.")


# ------------------------------------------------------------
# 2. READ THE COMPLETE FILE
# ------------------------------------------------------------

# "r" mode is used to read data from a file.

file = open("students.txt", "r")

data = file.read()

print(data)

file.close()
````

````python
# 4. readlines() - GET ALL LINES AS A LIST
file = open("students.txt", "r")
lines = file.readlines()
print(lines)
file.close()
````

````python
# 5. APPEND DATA TO AN EXISTING FILE
# "a" mode adds new content at the end.
# Existing content is not deleted.

file = open("students.txt", "a")
file.write("Sneha,88\n")
file.close()
print("New student added.")
````

````python

# 6. RECOMMENDED APPROACH - with open()
# Using "with open()" is preferred because Python
# automatically closes the file after the block is executed.

with open("students.txt", "r") as file:
    data = file.read()

print(data)

# No need to explicitly call file.close()

# 7. READ FILE AS A LIST OF LINES


with open("students.txt", "r") as file:
    lines = file.readlines()

# Remove the newline character from each line
lines = [line.strip() for line in lines]
print(lines)


# ------------------------------------------------------------
# 8. BASIC FILE OPERATION - CHECK WHETHER FILE EXISTS
# ------------------------------------------------------------

import os

if os.path.exists("students.txt"):
    print("File exists.")
else:
    print("File does not exist.")


# ------------------------------------------------------------
# QUICK SUMMARY
# ------------------------------------------------------------

# open("file.txt", "r")  -> Read
# open("file.txt", "w")  -> Write / Overwrite
# open("file.txt", "a")  -> Append
# file.read()            -> Read entire file
# file.readline()        -> Read one line
# file.readlines()       -> Read all lines as a list
# file.write()           -> Write data
# with open()            -> Recommended way to handle files
````
