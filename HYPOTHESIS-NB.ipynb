# Hypothesis Testing

Hypothesis testing is a statistical method used to determine whether
there is enough evidence in sample data to support a claim about a population.

For example:

A company claims that the average delivery time is 30 minutes.

We collect data from a sample of deliveries and ask:

"Does the sample provide enough evidence to conclude that the
average delivery time is different from 30 minutes?"

Hypothesis testing helps us make this decision mathematically.

The basic idea is:

Population → Sample → Evidence → Statistical Test → Decision
## Why Do We Need Hypothesis Testing?

We usually cannot collect data from the entire population.

Instead, we take a sample.

However, a sample can differ from the population simply because of
random sampling variation.

For example:

Population mean = 50

Sample 1 mean = 51
Sample 2 mean = 49
Sample 3 mean = 52

The question is:

"Is the difference meaningful, or could it have occurred simply
because of random chance?"

Hypothesis testing helps us answer this question.

# Null Hypothesis and Alternative Hypothesis

## Null Hypothesis (H₀)

The null hypothesis represents the default assumption.

It usually states that:

- There is no significant difference
- There is no effect
- There is no relationship
- A population parameter is equal to a specified value

Example:

A company claims that the average delivery time is 30 minutes.

H₀: μ = 30

---

## Alternative Hypothesis (H₁ or Hₐ)

The alternative hypothesis represents what we are trying to find
evidence for.

For example:

H₁: μ ≠ 30

This means the actual population mean may be different from 30 minutes.

The hypotheses are always statements about the POPULATION,
not just the sample.

````python
# Hypothesis testing example

# Population claim
claimed_mean = 30

# We will later compare our sample mean against this value
print("Claimed population mean:", claimed_mean)
````

Here, `30` represents the value specified in our **null hypothesis**:

$$H_0:\mu=30$$

The sample data will provide evidence that we will use to evaluate this claim.

## Creating Sample Data

Suppose a food-delivery company claims:

"The average delivery time is 30 minutes."

We collect delivery-time data from 15 randomly selected orders.

Our sample will be used to test the company's claim.

````python
import numpy as np

# Delivery time of 15 randomly selected orders
delivery_time = np.array([
    32, 28, 35, 31, 29,
    34, 30, 33, 27, 31,
    36, 29, 32, 30, 34
])

# Display the sample data
print("Delivery times:")
print(delivery_time)

# Calculate the sample mean
sample_mean = np.mean(delivery_time)

print("\nSample mean:", sample_mean)
````

# Significance Level (α)

The significance level determines how much evidence we require
before rejecting the null hypothesis.

The commonly used significance level is:

α = 0.05

This means we are willing to accept a 5% probability of making
a Type I error when the null hypothesis is actually true.

Other common values are:

α = 0.01
α = 0.05
α = 0.10

A smaller α means we require stronger evidence against H₀.

````python
# Set the significance level

alpha = 0.05

print("Significance level:", alpha)
print("Significance level (%):", alpha * 100, "%")
````

# p-value

The p-value measures how compatible the observed sample data is
with the null hypothesis.

More precisely:

The p-value is the probability, assuming the null hypothesis is true,
of obtaining a test statistic at least as extreme as the one observed.

A small p-value provides stronger evidence against H₀.

Decision rule:

If p-value ≤ α
    → Reject H₀

If p-value > α
    → Fail to reject H₀

Important:

We do NOT normally say "Accept H₀".

Instead, we say:

"Fail to reject H₀."

Why?

Because failing to find sufficient evidence against H₀ does not prove
that H₀ is definitely true.

# Test Statistic

A test statistic is a numerical value calculated from sample data
that measures how far our observed result is from what the null
hypothesis assumes.

For a one-sample z-test:

z = (x̄ - μ₀) / (σ / √n)

where:

x̄  = sample mean
μ₀  = mean assumed by H₀
σ   = population standard deviation
n   = sample size

A larger absolute test statistic means the sample result is farther
from the null hypothesis value.

# Decision Rule

The decision is based on the p-value and significance level.

If:

p-value ≤ α

we reject the null hypothesis.

If:

p-value > α

we fail to reject the null hypothesis.

Example:

α = 0.05

Case 1:
p-value = 0.02

0.02 < 0.05

→ Reject H₀


Case 2:
p-value = 0.18

0.18 > 0.05

→ Fail to reject H₀

# One-Tailed and Two-Tailed Tests

The type of test depends on the alternative hypothesis.

## Two-Tailed Test

Used when we want to know whether the parameter is simply different.

H₀: μ = 30

H₁: μ ≠ 30

The difference could be:

greater than 30 OR less than 30.

Therefore, both tails of the distribution are considered.


## Right-Tailed Test

Used when we are specifically testing whether the parameter is greater.

H₀: μ ≤ 30

H₁: μ > 30


## Left-Tailed Test

Used when we are specifically testing whether the parameter is smaller.

H₀: μ ≥ 30

H₁: μ < 30

## Choosing the Correct Test

Question:

"Is the average delivery time different from 30 minutes?"

Alternative:

H₁: μ ≠ 30

→ Two-tailed test


Question:

"Is the average delivery time greater than 30 minutes?"

Alternative:

H₁: μ > 30

→ Right-tailed test


Question:

"Is the average delivery time less than 30 minutes?"

Alternative:

H₁: μ < 30

→ Left-tailed test

# Type I and Type II Errors

Hypothesis testing involves decision-making under uncertainty.

Therefore, mistakes are possible.

## Type I Error

We reject H₀ when H₀ is actually true.

In simple words:

"We detected an effect that does not actually exist."

Type I error is represented by:

α

Example:

A company has NOT changed its process,
but our test concludes that the process has changed.


## Type II Error

We fail to reject H₀ when H₀ is actually false.

In simple words:

"We failed to detect a real effect."

Type II error is represented by:

β

Example:

A new training method really improves employee performance,
but our test fails to detect the improvement.

# Z-Test

A **Z-test** is a statistical hypothesis test used to determine whether a sample provides enough evidence to say that a population mean or proportion is significantly different from a hypothesized value.

### Simple Idea

Suppose a company claims:

> "The average delivery time is **30 minutes**."

So our hypothesis is:

$$H_0:\mu=30$$

We take a sample of deliveries and calculate its mean. The **Z-test** helps us decide whether the difference between the sample mean and 30 is large enough to be considered statistically significant.

### Z-Test Formula for a Mean

$$
Z=\frac{\bar{x}-\mu_0}{\sigma/\sqrt{n}}
$$

Where:

- $\bar{x}$ → Sample mean
- $\mu_0$ → Population mean assumed in the null hypothesis
- $\sigma$ → Population standard deviation
- $n$ → Sample size
- $Z$ → Z-test statistic

### Example

Suppose:

```text
Hypothesized mean = 30
Sample mean       = 32
Population SD     = 6
Sample size       = 36
```

Then:

$$
Z=\frac{32-30}{6/\sqrt{36}}
$$

$$
Z=\frac{2}{1}=2
$$

So, the **Z-test statistic is 2**.

We then use the Z-score to calculate the **p-value** and decide whether to reject or fail to reject $H_0$.

### When Do We Use a Z-Test?

A Z-test for a mean is typically appropriate when:

- The population standard deviation $\sigma$ is **known**.
- The population is approximately normal, or the sample is sufficiently large.
- We are testing a population mean against a hypothesized value.

If the Z-score is very far from `0`, the sample provides stronger evidence against the null hypothesis.

````python
import numpy as np
from statsmodels.stats.weightstats import ztest

# Sample battery-life data
battery_life = np.array([
    10.2, 9.8, 10.5, 10.1, 9.9,
    10.3, 10.0, 10.4, 9.7, 10.2
])

# Perform a one-sample z-test
# value=10 represents the population mean stated in H₀
z_stat, p_value = ztest(
    battery_life,
    value=10
)

print("Z-statistic:", z_stat)
print("p-value:", p_value)
````

````python
# Set significance level
alpha = 0.05

# Apply the decision rule
if p_value <= alpha:
    print("Reject H₀")
    print("There is sufficient evidence that the mean battery life differs from 10 hours.")
else:
    print("Fail to reject H₀")
    print("There is not sufficient evidence that the mean battery life differs from 10 hours.")
````

# One-Sample T-Test

A one-sample t-test is used to determine whether the mean of a sample
is significantly different from a hypothesized population mean.

It is commonly used when the population standard deviation is unknown.

This is very common in real-world situations.

Instead of using:

σ

we estimate the population standard deviation using:

s

The test statistic is:

t = (x̄ - μ₀) / (s / √n)

Degrees of freedom:

df = n - 1

````python
from scipy import stats
import numpy as np

# Sample exam scores
scores = np.array([
    72, 75, 68, 80, 74,
    77, 71, 69, 76, 73
])

# Hypothesized population mean
hypothesized_mean = 70

# Perform one-sample t-test
t_stat, p_value = stats.ttest_1samp(
    scores,
    popmean=hypothesized_mean
)

print("T-statistic:", t_stat)
print("p-value:", p_value)

# ttest_1samp() performs a one-sample t-test.

# It compares:
# - the sample mean of scores
# - with the hypothesized population mean of 70
#popmean=hypothesized_mean
# Compare the sample mean with the population mean specified in the null hypothesis.
````

````python
# Decision based on significance level

alpha = 0.05

if p_value <= alpha:
    print("Reject H₀")
else:
    print("Fail to reject H₀")
````

# Independent Two-Sample T-Test

An independent two-sample t-test compares the means of TWO
independent groups.

Example:

We want to determine whether two different training methods
produce different average test scores.

Group A → Students trained using Method A

Group B → Students trained using Method B

Hypotheses:

H₀: μ₁ = μ₂

H₁: μ₁ ≠ μ₂

The two groups should contain different/independent observations.

````python
from scipy import stats

# Test scores from two independent groups
group_A = [72, 75, 78, 70, 74, 77, 73, 76]
group_B = [81, 85, 79, 83, 88, 82, 84, 86]

# Perform independent two-sample t-test
t_stat, p_value = stats.ttest_ind(
    group_A,
    group_B
)

print("T-statistic:", t_stat)
print("p-value:", p_value)

#The t-statistic measures the difference between the two sample means relative to the estimated standard error of that difference.
````

````python
alpha = 0.05

if p_value <= alpha:
    print("Reject H₀")
    print("The two group means are significantly different.")
else:
    print("Fail to reject H₀")
    print("There is not enough evidence that the group means are different.")
````

# Paired T-Test

A paired t-test is used when the two sets of observations are related.

Common example:

Before-and-after measurements.

Example:

We measure employee performance:

Before training
After training

The same employees are measured twice.

Therefore, the observations are paired.

Hypotheses:

H₀: Mean difference = 0

H₁: Mean difference ≠ 0

# Chi-Square Test

Chi-square tests are commonly used with categorical data.

Two important applications are:

1. Chi-square goodness-of-fit test
2. Chi-square test of independence

The test of independence determines whether two categorical variables
are associated.

Example:

Is there a relationship between:

Gender and Product Preference?

Both variables are categorical.

H₀:
The two variables are independent.

H₁:
The two variables are associated.

The test compares:

Observed frequencies

with

Expected frequencies.

````python
import numpy as np
from scipy.stats import chi2_contingency

# Rows = Gender
# Columns = Product Preference

observed = np.array([
    [30, 20],
    [25, 25]
])

print("Observed frequencies:")
print(observed)

# Perform Chi-square test
chi2, p_value, degrees_of_freedom, expected = chi2_contingency(
    observed
)

print("\nChi-square statistic:", chi2)
print("Degrees of freedom:", degrees_of_freedom)
print("p-value:", p_value)
````

````python
print("Expected frequencies:")
print(expected)

alpha = 0.05

if p_value <= alpha:
    print("\nReject H₀")
    print("There is evidence of an association between the variables.")
else:
    print("\nFail to reject H₀")
    print("There is not enough evidence of an association.")
````

# ANOVA — Analysis of Variance

ANOVA is used to compare the means of THREE OR MORE groups.

Example:

A company wants to compare three training programs.

Program A → Test scores
Program B → Test scores
Program C → Test scores

Hypotheses:

H₀:
All population means are equal.

H₁:
At least one population mean is different.

ANOVA uses an F-statistic.

The basic idea is to compare:

Between-group variation

with

Within-group variation.

````python
from scipy import stats

# Test scores from three training programs

program_A = [72, 75, 78, 74, 70]
program_B = [81, 85, 79, 83, 88]
program_C = [76, 73, 75, 78, 74]

# Perform one-way ANOVA
f_stat, p_value = stats.f_oneway(
    program_A,
    program_B,
    program_C
)

print("F-statistic:", f_stat)
print("p-value:", p_value)
````

````python
alpha = 0.05

if p_value <= alpha:
    print("Reject H₀")
    print("At least one group mean is significantly different.")
else:
    print("Fail to reject H₀")
    print("There is not enough evidence that the group means differ.")
````

## Important: ANOVA Does Not Tell Us Which Groups Differ

Suppose ANOVA gives:

p-value = 0.01

We reject H₀.

This tells us:

"At least one group mean is different."

But it does NOT tell us:

A vs B
B vs C
A vs C

To determine which specific groups differ, we can perform
post-hoc tests such as Tukey's HSD.

Define the research question.

Step 2:
Define H₀.

Step 3:
Define H₁.

Step 4:
Choose the significance level α.

Step 5:
Select the appropriate statistical test.

Step 6:
Calculate the test statistic.

Step 7:
Calculate the p-value.

Step 8:
Compare p-value with α.

Step 9:
Make the statistical decision.

Step 10:
Interpret the result in the context of the problem.

--------------------------------------------------

Decision Rule:

If p-value ≤ α
→ Reject H₀

If p-value > α
→ Fail to reject H₀

--------------------------------------------------

Important:

Statistical significance does NOT automatically mean
practical/business significance.

A very small effect can become statistically significant
with a very large sample.

# Which Hypothesis Test Should I Use?

## One numerical sample vs known value

→ One-sample Z-test
→ One-sample T-test

## Two independent numerical groups

→ Independent two-sample T-test

Example:
Class A vs Class B

## Two related measurements

→ Paired T-test

Example:
Before vs After

## Three or more independent numerical groups

→ One-way ANOVA

Example:
Training A vs Training B vs Training C

## Two categorical variables

→ Chi-square Test of Independence

Example:
Gender vs Product Preference

## Categorical observed frequencies vs expected frequencies

→ Chi-square Goodness-of-Fit Test

## Relationship between numerical variables

→ Correlation / Regression

Example:
Experience vs Salary
